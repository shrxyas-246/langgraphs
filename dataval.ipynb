```python
from pydantic import BaseModel
#dataclass does not validate out  properly
```

```python
from typing import Optional 
class Employee(BaseModel):
    id : int 
    name: str
    department: str
    salary:Optional[float]=None #optional with default value 
    is_active:Optional[bool]=True
emp1=Employee(id=1,name="rey",department="IT")
print(emp1)
```

```python
from typing import List
from pydantic import Field
class Item(BaseModel):
    name :str=Field(min_length=2,description="provide a unique name")
    age : int =Field(default=18,description="AGE DEFAULTS TO 18")
it1=Item(name="robb",age=19)
print(it1.schema)
```

```python
class Person(BaseModel):
    Name : str
    Age : int
    city : str
person=Person(Name="Shre",Age=12,city="ind")
print(person)
type(person)
```
